```python
#🌟step 1 :                                                                   --- shravani
#Import the libraries & setup  required for data analysis and visualization.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8,6)
```

```python
#🌟step 1 :                                                                  ---- shravani
#Load dataset

df = pd.read_csv("cleaned_data.csv")

df.head()
```

```python
#🌟step 3:                                                                     ----vaishnavi
#Understand the Dataset
print(df.shape)

print(df.columns)

df.info()

df.describe()

df.describe(include="object")
```

```python
# 🌟step 4 : data cleaning                                                      ---vaishnavi

# count missing value
print(df.isnull().sum())

#Duplicate check
print(df.duplicated().sum())
```

```python
#Preview first 10 rows
df.head(10)
```

```python
#🌟Step 5:                                                                      ----vaishnavi
 #Descriptive Statistics


print("Average Followers:", df["followers"].mean())

print("Maximum Followers:", df["followers"].max())

print("Minimum Followers:", df["followers"].min())

print("Average Likes:", df["avg_likes"].mean())

print("Average Engagement:", df["60_day_eng_rate"].mean())
```

```python
#🌟Step 6: Top Influencer Analysis                                              ----suwarna

#Top 10 by followers
df.sort_values("followers",ascending=False).head(10)
```

```python
#🌟Top 10 by average likes                                                      ---suwarna
df.sort_values("avg_likes",ascending=False).head(10)
```

```python
#🌟Top 10 by influence score                                                    ---suwarna
df.sort_values("influence_score",ascending=False).head(10)
```

```python
#🌟Step 7: Country-wise Analysis                                               ---suwarna

#Country-wise distribution
df["country"].value_counts()
```

```python
#🌟average followers and average likes by country                               ---- suwarna
print(df.groupby("country")[["followers","avg_likes"]].mean())
```

```python
#Mean followers & mean avg likes
print(df["followers"].mean())
```

```python
print(df["avg_likes"].mean())
```

```python
#🌟Step 8: Correlation Analysis                                                 ---suwarna
df.corr(numeric_only=True)
```

```python
##🌟Step 9: Data Visualization                                                  ---nikita
#🌟Step 8: Pivot Table


pivot = df.pivot_table(
    values=["followers", "avg_likes"],
    index="country",
    aggfunc=["mean", "max", "count"]
)

pivot.columns = [
    "Mean Avg Likes",
    "Mean Followers",
    "Max Avg Likes",
    "Max Followers",
    "Count Avg Likes",
    "Count Followers"
]

print(pivot)
```

```python
#🌟Followers histogram                                                           ---nikita

df["followers"].hist(bins=20)

plt.title("Followers Distribution")

plt.xlabel("Followers")

plt.ylabel("Frequency")

plt.show()
```

```python
#🌟Avg likes histogram                                                          ---nikita

df["avg_likes"].hist(bins=20)
plt.title("Average Likes")
plt.show()
```

```python
#🌟Followers vs. likes scatter plot                                              ---nikita

plt.scatter(df["followers"], df["avg_likes"])

plt.xlabel("Followers")

plt.ylabel("Average Likes")

plt.title("Followers vs Average Likes")

plt.show()
```

```python
#🌟Followers boxplot                                                           ---nikita

df.boxplot(column="followers")

plt.title("Followers Box Plot")

plt.show()
```

```python
#🌟	Top 10 countries bar chart                                                  --- nikita

df["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top Countries")
plt.ylabel("Number of Influencers")
plt.show()
```

```python
#🌟Top 10 influencers bar chart                                                 --- nikita

top=df.sort_values("followers",ascending=False).head(10)

plt.bar(top["channel_info"],top["followers"])

plt.xticks(rotation=90)

plt.title("Top 10 Influencers")
plt.show()
```
